```python
import pandas as pd
import matplotlib.pyplot as plt

from src.analysis.market_analysis import kpi_summary, monthly_revenue, top_products

sales = pd.read_csv('data/processed/eligible_sales.csv', parse_dates=['InvoiceDate'])
rfm = pd.read_csv('data/processed/customer_rfm.csv')
kpi_summary(sales)
```

```python
monthly = monthly_revenue(sales)
ax = monthly.plot(x='Month', y='Revenue', kind='line', marker='o', figsize=(12, 5))
ax.set_title('Monthly Revenue')
ax.set_ylabel('Revenue')
plt.tight_layout()
plt.show()
```

```python
from src.models.customer_segmentation import segment_customers

segmented, model = segment_customers(rfm, cluster_count=5)
segmented.groupby('Segment').agg(
    Customers=('Customer ID', 'count'),
    Recency=('RecencyDays', 'mean'),
    Frequency=('Frequency', 'mean'),
    Monetary=('Monetary', 'mean'),
).sort_values('Monetary', ascending=False)
```
